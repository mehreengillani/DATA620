#### Import Libraries

```python
import networkx as nx
import matplotlib.pyplot as plt
```

#### Create Graph and add nodes

```python
G = nx.Graph()

# Add nodes (the 10 people in the network)
nodes = ["Amina", "Ali", "Ahmed", "Danish", "Edward", 
          "Fariha", "Gotham", "Haider", "Imran", "Jawad"]
G.add_nodes_from(nodes)

print("Nodes:", G.nodes())
print("Number of nodes:", G.number_of_nodes())
```

#### Add edges

```python
edges = [
    ("Amina", "Ali"),
    ("Amina", "Ahmed"),
    ("Amina", "Danish"),
    ("Amina", "Fariha"),
    ("Ali", "Danish"),
    ("Ali", "Edward"),
    ("Ali", "Gotham"),
    ("Ahmed", "Danish"),
    ("Ahmed", "Fariha"),
    ("Danish", "Edward"),
    ("Danish", "Fariha"),
    ("Danish", "Gotham"),
    ("Edward", "Gotham"),
    ("Fariha", "Gotham"),
    ("Fariha", "Haider"),
    ("Gotham", "Haider"),
    ("Haider", "Imran"),
    ("Imran", "Jawad"),
]
G.add_edges_from(edges)

print("Edges:", G.edges())
print("Number of edges:", G.number_of_edges())
```

### Draw the graph

```python
plt.figure(figsize=(10, 7))

# Fixed positions to roughly match the assignment's diagram
pos = {
    "Ahmed": (1, 4),
    "Amina": (0, 3),
    "Fariha": (2, 3),
    "Danish": (1, 2),
    "Ali": (0, 1),
    "Gotham": (2, 1),
    "Edward": (1, 0),
    "Haider": (3, 2),
    "Imran": (4, 2),
    "Jawad": (5, 2),
}

nx.draw(
    G, pos,
    with_labels=True,
    node_color="lightgray",
    node_size=1500,
    edge_color="purple",
    width=2,
    font_size=10,
    font_weight="bold",
)
plt.title("Graph", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()
```

### Basic graph analysis

```python
print("Degree of each node:")
for node, degree in G.degree():
    print(f"  {node}: {degree}")

print(f"\nGraph density: {nx.density(G):.3f}")
print(f"Is the graph connected? {nx.is_connected(G)}")
print(f"Average shortest path length: {nx.average_shortest_path_length(G):.3f}")
```
